```python
#import the gradio installed into the code
import gradio as gr

def track_finances(name, income, savings_goal, expenses_text):
    # Validate name and welcome
    if not name.strip():
        return "Please enter your name.", "", ""

    # Validate income range ($1,000 to $1,000,000)
    if income < 1000 or income > 1000000:
        return f"Hello {name}! Your yearly income must be between $1,000 and $1,000,000.", "", ""

    # Parse expenses from multiline text (format: Item: Price)
    expense_list = []
    total_monthly_expense = 0.0

    lines = expenses_text.strip().split("\n")
    for line in lines:
        if ":" in line:
            parts = line.split(":")
            item_name = parts[0].strip()
            try:
                item_price = float(parts[1].strip())
                expense_list.append((item_name, item_price))
                total_monthly_expense += item_price
            except ValueError:
                continue
        elif "," in line:
            parts = line.split(",")
            item_name = parts[0].strip()
            try:
                item_price = float(parts[1].strip())
                expense_list.append((item_name, item_price))
                total_monthly_expense += item_price
            except ValueError:
                continue

    # Rank expenses highest to lowest using sorting
    expense_list.sort(key=lambda x: x[1], reverse=True)

    # Calculations
    yearly_expense = total_monthly_expense * 12
    remaining_after_expenses = income - yearly_expense
    projected_savings_gap = savings_goal - remaining_after_expenses

    # Format ranked expenses list
    ranked_str = "### Expenses Ranked (Highest to Lowest):\n"
    if expense_list:
        for idx, (item, price) in enumerate(expense_list, 1):
            ranked_str += f"{idx}. **{item}**: ${price:,.2f} per month (${price * 12:,.2f} yearly)\n"
    else:
        ranked_str += "No valid expenses entered.\n"

    # Statistics and Goal Tracking
    stats_str = f"### Financial Statistics & Goal Tracking\n"
    stats_str += f"- **Yearly Income**: ${income:,.2f}\n"
    stats_str += f"- **Total Monthly Expenses**: ${total_monthly_expense:,.2f}\n"
    stats_str += f"- **Total Yearly Expenses**: ${yearly_expense:,.2f}\n"
    stats_str += f"- **Remaining Cashflow**: ${remaining_after_expenses:,.2f}\n"
    stats_str += f"- **Your Savings Goal**: ${savings_goal:,.2f}\n"

    if remaining_after_expenses >= savings_goal:
        stats_str += f"\n **Amazing job, {name}!** You are officially **on track** to crush your savings goal with a surplus of ${remaining_after_expenses - savings_goal:,.2f}!"
    else:
        stats_str += f"\n **Heads up, {name}.** You are falling short of your savings goal by ${abs(projected_savings_gap):,.2f}."

    # Personalized Suggestions
    suggestions_str = "### Personalized Suggestions\n"
    if remaining_after_expenses >= savings_goal:
        suggestions_str += "1. Consider moving your surplus into a high-yield savings account or investing it for long-term growth.\n"
        suggestions_str += "2. Reward yourself responsibly for staying disciplined with your budget!"
    else:
        if expense_list:
            suggestions_str += f"1. Review your highest monthly expense (**{expense_list[0][0]}** at ${expense_list[0][1]:,.2f}/mo) to see if you can trim it.\n"
        suggestions_str += f"2. Aim to reduce discretionary spending by at least 10-15% to bridge your ${abs(projected_savings_gap):,.2f} gap.\n"
        suggestions_str += "3. Look into automated savings transfers right after payday so you save first before spending."

    return ranked_str, stats_str, suggestions_str

# Build Gradio UI for Google Colab
with gr.Blocks(theme=gr.themes.Soft()) as demo:
    gr.Markdown("#  Personal Financial Assistant & Goal Tracker")
    gr.Markdown("Welcome! Enter your details below to analyze your income, track your expenses ranked by cost, and see how you're tracking toward your yearly savings goal.")

    with gr.Row():
        with gr.Column():
            name_input = gr.Textbox(label="Your Name", placeholder="e.g., Alex")
            income_input = gr.Number(label="Yearly Income ($)", value=50000, minimum=1000, maximum=1000000)
            savings_goal_input = gr.Number(label="Yearly Savings Goal ($)", value=10000)
            expenses_input = gr.Textbox(
                label="Monthly Expenses (Format: Item: Price per line)",
                placeholder="Rent: 1200\nGroceries: 450\nUtilities: 150\nEntertainment: 200",
                lines=5
            )
            submit_btn = gr.Button("Calculate My Finances", variant="primary")

        with gr.Column():
            ranked_output = gr.Markdown()
            stats_output = gr.Markdown()
            suggestions_output = gr.Markdown()

    submit_btn.click(
        fn=track_finances,
        inputs=[name_input, income_input, savings_goal_input, expenses_input],
        outputs=[ranked_output, stats_output, suggestions_output]
    )

demo.launch(quiet=True)
```
